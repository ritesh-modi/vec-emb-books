## Lesson 2 — make a vector

```python
# Activity: build the vector (3, 4) and look at it.
v = np.array([3, 4])

print("v =", v)
```

## Lesson 2 — read a vector's numbers
Counting starts at 0, so `v[0]` is the first number and `v[1]` is the second.

```python
# Activity: read the individual numbers of a vector.
v = np.array([3, 4])

print("first number,  v[0]:", v[0])
print("second number, v[1]:", v[1])
```

## Lesson 3 — add two vectors
Adding matches up the numbers: first-with-first, second-with-second.

```python
# Activity: add two vectors.
a = np.array([4, 1])
b = np.array([1, 3])

result = a + b
print("a + b =", result)        # -> [5 4]
```

## Lesson 3 — subtract two vectors
Subtracting gives the arrow that takes you from one point to the other.

```python
# Activity: subtract two vectors.
a = np.array([5, 7])
b = np.array([2, 3])

result = a - b
print("a - b =", result)        # -> [3 4]
```

## Lesson 3 — stretch, shrink, or flip a vector
Multiplying by a number changes the length; a negative number flips the direction.

```python
# Activity: multiply a vector by a number.
a = np.array([4, 1])

print("2 * a   =", 2 * a)       # stretch -> [8 2]
print("0.5 * a =", 0.5 * a)     # shrink  -> [2.  0.5]
print("-1 * a  =", -1 * a)      # flip    -> [-4 -1]
```

## Lesson 4 — the length (magnitude) of a vector
`np.linalg.norm(v)` returns the straight-line length of `v`.
(`linalg` is NumPy's linear-algebra toolbox.)

```python
# Activity: measure the length of one vector.
v = np.array([3, 4])

length = np.linalg.norm(v)
print("length of (3, 4):", length)      # -> 5.0
```

## Lesson 4 — a length that is a long decimal
Some lengths are not whole numbers. Round to 2 places when you need to read them.

```python
# Activity: measure a length and round it.
v = np.array([2, 3])

length = np.linalg.norm(v)
print("length of (2, 3):", round(float(length), 2))   # -> 3.61
```

## Lesson 4 — distance between two points
Subtract to get the arrow between the points, then measure that arrow's length.

```python
# Activity: find the distance between two points.
p = np.array([1, 1])
q = np.array([4, 5])

arrow = q - p                            # the arrow from p to q
distance = np.linalg.norm(arrow)
print("distance from (1,1) to (4,5):", distance)   # -> 5.0
```
